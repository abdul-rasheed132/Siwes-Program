# WEEK 6: Data Visualization with Matplotlib & Seaborn

Monday- Introduction to matplotlib

```python
# Cell 1: Basic Line Plot
import matplotlib.pyplot as plt

# Simple data
x = [1, 2, 3, 4, 5]
y = [10, 25, 15, 30, 20]

plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

```python
# Cell 2: Bar Chart
import matplotlib.pyplot as plt

products = ['Laptop', 'Phone', 'Tablet', 'Monitor']
sales = [1200, 800, 500, 300]

plt.bar(products, sales, color='steelblue')
plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.show()
```

```python
# Cell 3: Customizing Plots
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [10, 25, 15, 30, 20]
y2 = [5, 15, 10, 25, 15]

plt.plot(x, y1, marker='o', linestyle='-', color='blue', label='Product A')
plt.plot(x, y2, marker='s', linestyle='--', color='red', label='Product B')
plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
```

TUESDAY - Advanced Matplotlib Customization


```python
# Cell 1: Subplots
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [10, 25, 15, 30, 20]
y2 = [5, 15, 10, 25, 15]

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(x, y1, color='blue')
axes[0].set_title("Product A")

axes[1].plot(x, y2, color='red')
axes[1].set_title("Product B")

plt.tight_layout()
plt.show()
```

```python
# Cell 2: Pie Chart and Scatter Plot
import matplotlib.pyplot as plt

# Pie Chart
sizes = [30, 25, 20, 25]
labels = ['Laptop', 'Phone', 'Tablet', 'Monitor']
plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title("Sales Distribution")
plt.show()

# Scatter Plot
x = [1, 2, 3, 4, 5]
y = [10, 25, 15, 30, 20]
plt.scatter(x, y, color='green', s=100)
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
# Cell 3: Saving Figures
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 25, 15, 30, 20]

plt.figure(figsize=(8, 5))
plt.plot(x, y, marker='o')
plt.title("Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

# Save the figure
plt.savefig('sales_trend.png', dpi=300, bbox_inches='tight')
plt.show()
print("Figure saved as sales_trend.png")
```

TUESDAY - Heatmaps


```python
# Cell 1: Basic Heatmap
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset("iris")
correlation = iris.corr(numeric_only=True)

sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Cell 2: Customized Heatmap
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

iris = sns.load_dataset("iris")
correlation = iris.corr(numeric_only=True)

# Customized heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation,
    annot=True,
    cmap='RdYlGn',
    center=0,
    fmt='.2f',
    linewidths=0.5,
    square=True
)
plt.title("Correlation Heatmap (Customized)")
plt.show()
```

```python
# Cell 3: Heatmap with Mask
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

iris = sns.load_dataset("iris")
correlation = iris.corr(numeric_only=True)

# Create mask for upper triangle
mask = np.triu(np.ones_like(correlation, dtype=bool))

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation,
    mask=mask,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
plt.title("Lower Triangle Correlation Heatmap")
plt.show()
```

WEDNESDAY - Subplots & Complex Figures


```python
# Cell 1: Creating Subplots
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title("Sine")

axes[0, 1].plot(x, np.cos(x), color='orange')
axes[0, 1].set_title("Cosine")

axes[1, 0].plot(x, np.tan(x), color='green')
axes[1, 0].set_title("Tangent")

axes[1, 1].plot(x, x**2, color='red')
axes[1, 1].set_title("Square")

plt.tight_layout()
plt.show()
```

```python
# Cell 2: Subplots with Shared Axes
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

fig, axes = plt.subplots(2, 1, figsize=(8, 6), sharex=True)

axes[0].plot(x, np.sin(x), color='blue')
axes[0].set_ylabel("Sin(x)")
axes[0].grid(True)

axes[1].plot(x, np.cos(x), color='red')
axes[1].set_ylabel("Cos(x)")
axes[1].set_xlabel("X values")
axes[1].grid(True)

plt.suptitle("Sine and Cosine Waves")
plt.tight_layout()
plt.show()
```

```python
# Cell 3: GridSpec for Complex Layouts
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

fig = plt.figure(figsize=(10, 6))
gs = gridspec.GridSpec(2, 3, figure=fig)

ax1 = fig.add_subplot(gs[0, :])
ax1.set_title("Top Row (Full Width)")

ax2 = fig.add_subplot(gs[1, 0])
ax2.set_title("Bottom Left")

ax3 = fig.add_subplot(gs[1, 1:])
ax3.set_title("Bottom Right (Spanning 2 Columns)")

plt.tight_layout()
plt.show()
```

THURSDAY - Seaborn Themes & Color Palettes


```python
# Cell 1: Seaborn Themes
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

# Available themes
themes = ['darkgrid', 'whitegrid', 'dark', 'white', 'ticks']
print("Available themes:", themes)

# Apply darkgrid
sns.set_theme(style="darkgrid")
sns.barplot(x='day', y='total_bill', data=tips)
plt.title("Theme: Darkgrid")
plt.show()
```

```python
# Cell 2: Comparing Themes
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, style in zip(axes, ['whitegrid', 'dark', 'ticks']):
    sns.set_theme(style=style)
    sns.barplot(x='day', y='total_bill', data=tips, ax=ax)
    ax.set_title(f"Theme: {style}")

plt.tight_layout()
plt.show()
```

```python
# Cell 3: Color Palettes
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

# Different palettes
palettes = ['deep', 'muted', 'bright', 'pastel', 'dark']

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.set_theme(style="whitegrid")

sns.barplot(x='day', y='total_bill', data=tips, palette='bright', ax=axes[0])
axes[0].set_title("Bright Palette")

sns.barplot(x='day', y='total_bill', data=tips, palette='pastel', ax=axes[1])
axes[1].set_title("Pastel Palette")

sns.barplot(x='day', y='total_bill', data=tips, palette='dark', ax=axes[2])
axes[2].set_title("Dark Palette")

plt.tight_layout()
plt.show()
```

FRIDAY - Time Series Visualization

```python
# Cell 1: Basic Time Series Plot
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Create date range
dates = pd.date_range(start='2026-01-01', periods=12, freq='ME')
sales = [100, 120, 115, 140, 160, 155, 170, 180, 175, 190, 200, 210]

df = pd.DataFrame({'Date': dates, 'Sales': sales})
print(df)

plt.plot(df['Date'], df['Sales'], marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Cell 2: Time Series with Multiple Lines
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

dates = pd.date_range(start='2026-01-01', periods=12, freq='ME')
product_a = [100, 120, 115, 140, 160, 155, 170, 180, 175, 190, 200, 210]
product_b = [80, 90, 95, 100, 110, 105, 120, 125, 130, 135, 140, 150]

plt.figure(figsize=(10, 5))
plt.plot(dates, product_a, marker='o', label='Product A')
plt.plot(dates, product_b, marker='s', label='Product B')
plt.title("Sales Trend Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Cell 3: Time Series with Seaborn
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create data
dates = pd.date_range(start='2026-01-01', periods=12, freq='ME')
df = pd.DataFrame({
    'Date': dates,
    'Sales': [100, 120, 115, 140, 160, 155, 170, 180, 175, 190, 200, 210],
    'Product': ['A'] * 12
})

df2 = pd.DataFrame({
    'Date': dates,
    'Sales': [80, 90, 95, 100, 110, 105, 120, 125, 130, 135, 140, 150],
    'Product': ['B'] * 12
})

combined = pd.concat([df, df2])
print(combined)

sns.lineplot(x='Date', y='Sales', hue='Product', data=combined, marker='o')
plt.title("Sales Trend by Product")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
