MONDAY - Introduction to groupby()

```python
# Cell 1: Basic groupby()
import pandas as pd
import io

csv_data = """Name,Department,Salary,City
Alice,IT,50000,Lagos
Bob,HR,60000,Abuja
Carol,IT,55000,Ibadan
David,Finance,45000,Lagos
Eve,HR,70000,Lagos
Frank,IT,65000,Abuja"""
df = pd.read_csv(io.StringIO(csv_data))
print("Original Data:")
print(df)

# Group by Department
groups = df.groupby('Department')
print(f"\nGroups: {groups.groups.keys()}")
print(f"Group sizes:\n{groups.size()}")
```

```python
# Cell 2: Group Statistics
import pandas as pd
import io

csv_data = """Name,Department,Salary,City
Alice,IT,50000,Lagos
Bob,HR,60000,Abuja
Carol,IT,55000,Ibadan
David,Finance,45000,Lagos
Eve,HR,70000,Lagos
Frank,IT,65000,Abuja"""
df = pd.read_csv(io.StringIO(csv_data))
print(df)
# Average salary by department
print("Average Salary by Department:")
print(df.groupby('Department')['Salary'].mean())

# Count by department
print("\nEmployee Count by Department:")
print(df.groupby('Department')['Name'].count())
```

```python
# Cell 3: Multiple Group Columns
import pandas as pd
import io

csv_data = """Name,Department,Salary,City
Alice,IT,50000,Lagos
Bob,HR,60000,Abuja
Carol,IT,55000,Ibadan
David,Finance,45000,Lagos
Eve,HR,70000,Lagos
Frank,IT,65000,Abuja"""
df = pd.read_csv(io.StringIO(csv_data))

# Group by Department and City
print("Group by Department and City:")
print(df.groupby(['Department', 'City'])['Salary'].mean())

# Get group for specific department
it_group = df.groupby('Department').get_group('IT')
print(f"\nIT Department:\n{it_group}")
```

TUESDAY - Aggregate Functions on Groups


```python
# Cell 1: Basic Aggregations
import pandas as pd
import io

csv_data = """Name,Department,Salary,Age
Alice,IT,50000,25
Bob,HR,60000,30
Carol,IT,55000,28
David,Finance,45000,22
Eve,HR,70000,35
Frank,IT,65000,32"""
df = pd.read_csv(io.StringIO(csv_data))

# Multiple aggregations
print("Salary Statistics by Department:")
print(df.groupby('Department')['Salary'].agg(['mean', 'sum', 'count', 'min', 'max']))
```

```python
# Cell 2: Aggregating Multiple Columns
import pandas as pd
import io

csv_data = """Name,Department,Salary,Age
Alice,IT,50000,25
Bob,HR,60000,30
Carol,IT,55000,28
David,Finance,45000,22
Eve,HR,70000,35
Frank,IT,65000,32"""
df = pd.read_csv(io.StringIO(csv_data))

# Aggregate different columns
print("Aggregations by Department:")
print(df.groupby('Department').agg({
    'Salary': ['mean', 'sum'],
    'Age': ['mean', 'max']
}))
```

```python
# Cell 3: Custom Aggregations with Named Functions
import pandas as pd
import io

csv_data = """Name,Department,Salary,Age
Alice,IT,50000,25
Bob,HR,60000,30
Carol,IT,55000,28
David,Finance,45000,22
Eve,HR,70000,35
Frank,IT,65000,32"""
df = pd.read_csv(io.StringIO(csv_data))

# Using named aggregation
result = df.groupby('Department').agg(
    Avg_Salary=('Salary', 'mean'),
    Total_Salary=('Salary', 'sum'),
    Max_Age=('Age', 'max'),
    Employee_Count=('Name', 'count')
)
print("Named Aggregations:")
print(result)
```

WEDNESDAY - Advanced Aggregation and Pivot Tables

```python
# Cell 1: Multiple Aggregations with .agg()
import pandas as pd
import io

csv_data = """Name,Department,Salary,Age,City
Alice,IT,50000,25,Lagos
Bob,HR,60000,30,Abuja
Carol,IT,55000,28,Ibadan
David,Finance,45000,22,Lagos
Eve,HR,70000,35,Lagos
Frank,IT,65000,32,Abuja"""
df = pd.read_csv(io.StringIO(csv_data))

# Multiple functions on single column
print("Salary Aggregations:")
print(df.groupby('Department')['Salary'].agg(['mean', 'median', 'std']))

# Different functions on different columns
print("\nDifferent Aggregations:")
print(df.groupby('Department').agg({
    'Salary': ['mean', 'sum'],
    'Age': ['min', 'max']
}))
```


THURSDAY - Data Transformation with apply(

```python
# Cell 1: Using apply() on Columns
import pandas as pd
import io

csv_data = """Name,Salary,Age
Alice,50000,25
Bob,60000,30
Carol,55000,28
David,45000,22"""
df = pd.read_csv(io.StringIO(csv_data))
print("Original:")
print(df)

# Apply custom function to column
df['Salary_K'] = df['Salary'].apply(lambda x: x / 1000)
print("\nAfter apply (Salary in thousands):")
print(df)
```

```python
# Cell 2: Using apply() on Rows
import pandas as pd
import io

csv_data = """Name,Salary,Age
Alice,50000,25
Bob,60000,30
Carol,55000,28
David,45000,22"""
df = pd.read_csv(io.StringIO(csv_data))

# Apply function to rows
def categorize(row):
    if row['Salary'] > 55000:
        return 'High'
    elif row['Salary'] > 48000:
        return 'Medium'
    else:
        return 'Low'

df['Category'] = df.apply(categorize, axis=1)
print("After row apply:")
print(df)
```

```python
# Cell 3: applymap() for Element-wise Operations
import pandas as pd
import io

csv_data = """Name,Salary,Age
Alice,50000,25
Bob,60000,30
Carol,55000,28
David,45000,22"""
df = pd.read_csv(io.StringIO(csv_data))

# Select numeric columns
numeric_df = df[['Salary', 'Age']]
print("Numeric columns:")
print(numeric_df)

# Apply to all elements
df_doubled = numeric_df.applymap(lambda x: x * 2)
print("\nAfter applymap (doubled):")
print(df_doubled)
```

```python
# Cell 1: Sales Dataset Analysis
import pandas as pd
import io

sales_data = """Date,Product,Region,Salesperson,Units,Price
2026-01-15,Laptop,Lagos,Alice,5,1200
2026-01-16,Phone,Abuja,Bob,10,800
2026-01-17,Tablet,Lagos,Carol,8,500
2026-01-18,Laptop,Abuja,David,3,1200
2026-01-19,Phone,Lagos,Eve,15,800
2026-01-20,Tablet,Abuja,Frank,12,500
2026-01-21,Laptop,Ibadan,Alice,7,1200
2026-01-22,Phone,Ibadan,Bob,9,800"""
df = pd.read_csv(io.StringIO(sales_data))

# Calculate total sales
df['Total'] = df['Units'] * df['Price']
print("Sales Data with Total:")
print(df)
```

```python
# Cell 2: Finding Top Products and Regions
import pandas as pd
import io

sales_data = """Date,Product,Region,Salesperson,Units,Price
2026-01-15,Laptop,Lagos,Alice,5,1200
2026-01-16,Phone,Abuja,Bob,10,800
2026-01-17,Tablet,Lagos,Carol,8,500
2026-01-18,Laptop,Abuja,David,3,1200
2026-01-19,Phone,Lagos,Eve,15,800
2026-01-20,Tablet,Abuja,Frank,12,500
2026-01-21,Laptop,Ibadan,Alice,7,1200
2026-01-22,Phone,Ibadan,Bob,9,800"""
df = pd.read_csv(io.StringIO(sales_data))
df['Total'] = df['Units'] * df['Price']

# Top products by revenue
print("Revenue by Product:")
print(df.groupby('Product')['Total'].sum().sort_values(ascending=False))

# Top regions by revenue
print("\nRevenue by Region:")
print(df.groupby('Region')['Total'].sum().sort_values(ascending=False))
```

```python
# Cell 3: Complete Sales Summary
import pandas as pd
import io

sales_data = """Date,Product,Region,Salesperson,Units,Price
2026-01-15,Laptop,Lagos,Alice,5,1200
2026-01-16,Phone,Abuja,Bob,10,800
2026-01-17,Tablet,Lagos,Carol,8,500
2026-01-18,Laptop,Abuja,David,3,1200
2026-01-19,Phone,Lagos,Eve,15,800
2026-01-20,Tablet,Abuja,Frank,12,500
2026-01-21,Laptop,Ibadan,Alice,7,1200
2026-01-22,Phone,Ibadan,Bob,9,800"""
df = pd.read_csv(io.StringIO(sales_data))
df['Total'] = df['Units'] * df['Price']

# Complete summary
summary = df.groupby('Product').agg(
    Total_Revenue=('Total', 'sum'),
    Units_Sold=('Units', 'sum'),
    Avg_Price=('Price', 'mean'),
    Transactions=('Total', 'count')
).sort_values('Total_Revenue', ascending=False)

print("Complete Product Summary:")
print(summary)
```
